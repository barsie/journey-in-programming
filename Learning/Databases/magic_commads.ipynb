2. %pwd: prints working directory

5. %who: list all variables in the name space
6. %whos: provide more detail information of the variable in the name spaces
7. %matplotlib inline: makes matplotlibs plots appear within the notebook
8. %timeit: time the execution of a single statment
9. %lsmagic: list all available line magics
