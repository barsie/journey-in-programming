```python
class TreeNode:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def __str__(self):
        return str(self.val)
```

```python
A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(14)
G = TreeNode(19)

A.left = B
A.right = C
B.left = D
B.right = E
D.left = F
D.right = G

print(A)
```

```python
# Recursive pre-order Traversal (DSF) Time O(n), Space O(n)

def pre_order(node):
    if not node: return

    print(node)
    pre_order(node.left)
    pre_order(node.right)

pre_order(A)
```

```python
# Recursive in-order Tranversal (DSF) Time O(n), Space(n)

def in_order(node):
    if not node: return

    pre_order(node.left)
    print(node)
    pre_order(node.right)

in_order(A)
```

```python
def post_order(node):
    if not node: return

    pre_order(node.left)
    pre_order(node.right)
    print(node)

post_order(A)
```

```python
""" Iterative Pre-order Transversal
    - Uses a stack algorithm
    - And them processes all the requirement as:
        - process node
        - go to left
        - go to right
    All using a while 
 """
def pre_order_iterative(node):
    stk = [node]

    while stk:
        node = stk.pop()
        print(node)
        if node.left: stk.append(node.left)
        if node.right: stk.append(node.right)

pre_order_iterative(A)
```

```python
"""
    The BSF: Brest First Search, read the three with a QUEUE FIFO approach
    with Time complexity: O(n) and Space O(n)
    The result of BSF is a sorted array

    Below is an implementation with Level of Order Traversal
"""

from collections import deque

def level_order(node):
    q = deque()
    q.append(node)

    while q:
        node = q.popleft()
        print(node)
        if node.left: q.append(node.left)
        if node.right: q.append(node.right)

level_order(A)
```
