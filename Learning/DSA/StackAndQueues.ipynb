```python
# Stack Implementation using List and related built-in methods.

class Stack: 
    def __init__(self):
        self.items = []
    
    # create push
    def push(self, items):
        self.items.append(items)

    # Create pop
    def pop(self):
        self.items.pop()

    # Peek and item
    def peek(self):
        return self.items[-1]
    
    def __str__(self):
        return str(self.items)
```

```python
# create a main path: passing all Stack class attr to a variable.
if __name__ == "__main__":
    s = Stack()
    print(s)

    s.push(23)
    s.push(0)
    print(s)
    s.push(-12)
    s.push(100)
    print(s)

    print(s.pop())
    print(s)
```
