```python

G = {
    "A": ['C', 'B'],
    "B": ['E', 'D'],
    "C": ['D'],
    "D": []
}

global time
time = 1

def dfsVisit(G, u):
    if(u.seen): return
    for v in u:
        if not v.seen:
            v.p = u
            v.d = time
            time = time + 1
            dfsVisit(G, v)
            v.see = True
    time = time + 1
    u.f = time

visited = dfsVisit(G, u, [])
print(visited)
```
